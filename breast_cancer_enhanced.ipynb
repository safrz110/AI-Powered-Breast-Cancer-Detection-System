```python
# Imports
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

```python
# Drop irrelevant columns (same as original)
df.drop(columns=['id', 'Unnamed: 32'], inplace=True, errors='ignore')
df.head()
```

```python
X = df.drop('diagnosis', axis=1)
Y = df['diagnosis']

num_cols = X.select_dtypes(include=["int64", "float64"]).columns
cat_cols = X.select_dtypes(include=["object"]).columns

from sklearn.preprocessing import OneHotEncoder, StandardScaler

num_transformer = StandardScaler()
cat_transformer = OneHotEncoder(handle_unknown='ignore')


from sklearn.compose import ColumnTransformer

preprocessor = ColumnTransformer([
    ("Num", num_transformer, num_cols),
    ("Cat", cat_transformer, cat_cols)
])

from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.2, random_state=42, stratify=Y 
)

print(f"Training set shape : {X_train.shape}")
print(f"Testing  set shape : {X_test.shape}")
```

```python
from sklearn.pipeline import Pipeline
from sklearn.neighbors import KNeighborsClassifier

baseline_pipe = Pipeline([
    ("Pre", preprocessor),
    ("KNN", KNeighborsClassifier(n_neighbors=5))
])

baseline_pipe.fit(X_train, Y_train)
Y_pred_baseline = baseline_pipe.predict(X_test)

from sklearn.metrics import accuracy_score, classification_report

print("Baseline KNN (k=5) ")
print(f"Accuracy : {accuracy_score(Y_test, Y_pred_baseline):.4f}")
print(classification_report(Y_test, Y_pred_baseline))
```

```python
# ── Cross-Validation on the baseline KNN pipeline 
from sklearn.model_selection import StratifiedKFold, cross_val_score

cv_strategy = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
cv_scores = cross_val_score(
    baseline_pipe,
    X, Y,                  
    cv=cv_strategy,
    scoring='accuracy'
)

print("Cross-Validation Accuracy scores per fold:", np.round(cv_scores, 4))
print(f"Mean CV Accuracy : {cv_scores.mean():.4f}")
print(f"Std  CV Accuracy : {cv_scores.std():.4f}  (lower = more stable model)")

# Visualise the fold scores
plt.figure(figsize=(7, 4))
plt.bar(range(1, 6), cv_scores, color='steelblue', edgecolor='black')
plt.axhline(cv_scores.mean(), color='red', linestyle='--', label=f'Mean = {cv_scores.mean():.4f}')
plt.xlabel('Fold'); plt.ylabel('Accuracy')
plt.title('5-Fold Cross-Validation — Baseline KNN')
plt.legend(); plt.tight_layout(); plt.show()
```

```python
# ── GridSearchCV on KNN pipeline 
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV

knn_param_grid = {
    'KNN__n_neighbors': [3, 5, 7, 9, 11, 15],          
    'KNN__metric':      ['euclidean', 'manhattan'],      
    'KNN__weights':     ['uniform', 'distance']          
}


grid_search_knn = GridSearchCV(
    estimator=Pipeline([
        ("Pre", preprocessor),
        ("KNN", KNeighborsClassifier())
    ]),
    param_grid=knn_param_grid,
    cv=cv_strategy,          
    scoring='accuracy',
    n_jobs=-1,               
    refit=True,              
    verbose=1
)

grid_search_knn.fit(X_train, Y_train)

print("Best KNN parameters :", grid_search_knn.best_params_)
print(f"Best CV accuracy    : {grid_search_knn.best_score_:.4f}")

# Evaluate the tuned KNN on the held-out test set
best_knn_pred = grid_search_knn.predict(X_test)
print("\n=== Tuned KNN — Test Set ===")
print(f"Accuracy : {accuracy_score(Y_test, best_knn_pred):.4f}")
print(classification_report(Y_test, best_knn_pred))
```

```python
# Visualise the K-sweep (now with best metric/weights)
best_metric  = grid_search_knn.best_params_['KNN__metric']
best_weights = grid_search_knn.best_params_['KNN__weights']

scores_k = []
for k in range(1, 25):
    pipe_k = Pipeline([
        ("Pre", preprocessor),
        ("KNN", KNeighborsClassifier(n_neighbors=k, metric=best_metric, weights=best_weights))
    ])
    pipe_k.fit(X_train, Y_train)
    scores_k.append(accuracy_score(Y_test, pipe_k.predict(X_test)))

plt.plot(range(1, 25), scores_k, marker='o')
plt.xlabel('K'); plt.ylabel('Accuracy')
plt.title(f'KNN K-sweep (metric={best_metric}, weights={best_weights})')
plt.tight_layout(); plt.show()
```

```python
# ── Random Forest with GridSearchCV ──────────────────────────────────────────
from sklearn.ensemble import (RandomForestClassifier,GradientBoostingClassifier,)

rf_param_grid = {
    'RF__n_estimators':      [100, 200],          
    'RF__max_depth':         [None, 10, 20],       
    'RF__min_samples_split': [2, 5],               
    'RF__max_features':      ['sqrt', 'log2']     
}

rf_pipe = Pipeline([
    ("Pre", preprocessor),
    ("RF", RandomForestClassifier(random_state=42))
])

grid_search_rf = GridSearchCV(
    rf_pipe,
    rf_param_grid,
    cv=cv_strategy,
    scoring='accuracy',
    n_jobs=-1,
    refit=True,
    verbose=1
)

grid_search_rf.fit(X_train, Y_train)

print("Best RF parameters :", grid_search_rf.best_params_)
print(f"Best CV accuracy   : {grid_search_rf.best_score_:.4f}")

rf_pred = grid_search_rf.predict(X_test)
print("\nTuned Random Forest — Test Set ")
print(f"Accuracy : {accuracy_score(Y_test, rf_pred):.4f}")
print(classification_report(Y_test, rf_pred))
```

```python
# Feature importance from the best Random Forest
best_rf_model = grid_search_rf.best_estimator_.named_steps['RF']

# Get feature names after ColumnTransformer
feat_names = list(num_cols) + (
    list(grid_search_rf.best_estimator_.named_steps['Pre']
         .named_transformers_['Cat']
         .get_feature_names_out(cat_cols)) if len(cat_cols) > 0 else []
)

importances = pd.Series(best_rf_model.feature_importances_, index=feat_names)
importances.sort_values().tail(15).plot(kind='barh', figsize=(8, 6))
plt.title('Top 15 Feature Importances — Random Forest')
plt.tight_layout(); plt.show()
```

```python
# ── Gradient Boosting with RandomizedSearchCV 
from scipy.stats import uniform

gb_param_dist = {
    'GB__n_estimators':  [100, 200, 300],
    'GB__learning_rate': uniform(0.01, 0.29),   
    'GB__max_depth':     [3, 4, 5],
    'GB__subsample':     uniform(0.6, 0.4)      
}

gb_pipe = Pipeline([
    ("Pre", preprocessor),
    ("GB", GradientBoostingClassifier(random_state=42))
])

rand_search_gb = RandomizedSearchCV(
    gb_pipe,
    gb_param_dist,
    n_iter=20,
    cv=cv_strategy,
    scoring='accuracy',
    n_jobs=-1,
    refit=True,
    random_state=42,
    verbose=1
)

rand_search_gb.fit(X_train, Y_train)

print("Best GB parameters :", rand_search_gb.best_params_)
print(f"Best CV accuracy   : {rand_search_gb.best_score_:.4f}")

gb_pred = rand_search_gb.predict(X_test)
print("\n Tuned Gradient Boosting — Test Set")
print(f"Accuracy : {accuracy_score(Y_test, gb_pred):.4f}")
print(classification_report(Y_test, gb_pred))
```

```python
# Collect all models and their predictions
models = {
    'Baseline KNN (k=5)'     : (baseline_pipe,        Y_pred_baseline),
    'Tuned KNN'              : (grid_search_knn,       best_knn_pred),
    'Tuned Random Forest'    : (grid_search_rf,        rf_pred),
    'Tuned Gradient Boosting': (rand_search_gb,        gb_pred)
}

results = []
for name, (model, preds) in models.items():
    # roc_auc needs probability scores
    try:
        proba = model.predict_proba(X_test)[:, 1]
        auc = roc_auc_score(Y_test, proba)
    except Exception:
        auc = float('nan')
    results.append({
        'Model'   : name,
        'Accuracy': accuracy_score(Y_test, preds),
        'ROC-AUC' : auc
    })

results_df = pd.DataFrame(results).sort_values('Accuracy', ascending=False)
print(results_df.to_string(index=False))
```

```python
# Bar chart — Accuracy comparison
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].barh(results_df['Model'], results_df['Accuracy'], color='steelblue', edgecolor='black')
axes[0].set_xlim(0.88, 1.01)
axes[0].set_xlabel('Accuracy'); axes[0].set_title('Model Accuracy Comparison')
for i, v in enumerate(results_df['Accuracy']):
    axes[0].text(v + 0.001, i, f'{v:.4f}', va='center')

axes[1].barh(results_df['Model'], results_df['ROC-AUC'], color='coral', edgecolor='black')
axes[1].set_xlim(0.88, 1.01)
axes[1].set_xlabel('ROC-AUC'); axes[1].set_title('Model ROC-AUC Comparison')
for i, v in enumerate(results_df['ROC-AUC']):
    axes[1].text(v + 0.001, i, f'{v:.4f}', va='center')

plt.tight_layout(); plt.show()
```

```python
# Confusion matrices for all models
from sklearn.metrics import (ConfusionMatrixDisplay,roc_auc_score,roc_curve)
fig, axes = plt.subplots(2, 3, figsize=(16, 10))
axes = axes.flatten()

for ax, (name, (model, preds)) in zip(axes, models.items()):
    ConfusionMatrixDisplay.from_predictions(
        Y_test, preds,
        display_labels=['Benign', 'Malignant'],
        ax=ax, colorbar=False, cmap='Blues'
    )
    ax.set_title(name, fontsize=9)

plt.suptitle('Confusion Matrices — All Models', fontsize=13, y=1.02)
plt.tight_layout(); plt.show()
```

```python
# ROC curves
plt.figure(figsize=(9, 6))
for name, (model, _) in models.items():
    try:
        proba = model.predict_proba(X_test)[:, 1]
        fpr, tpr, _ = roc_curve(Y_test, proba)
        auc = roc_auc_score(Y_test, proba)
        plt.plot(fpr, tpr, label=f'{name} (AUC={auc:.3f})')
    except Exception:
        pass

plt.plot([0,1],[0,1],'k--', label='Random classifier')
plt.xlabel('False Positive Rate'); plt.ylabel('True Positive Rate')
plt.title('ROC Curves — All Models')
plt.legend(fontsize=8); plt.tight_layout(); plt.show()
```
